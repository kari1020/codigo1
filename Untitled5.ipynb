```python
CATALOGO = {
    "001": ("Café americano", 4500, 0.19),
    "002": ("Capuchino", 6000, 0.19),
    "003": ("Croissant jamón y queso", 7500, 0.19),
    "004": ("Pan artesanal", 5000, 0.05),
    "005": ("Jugo natural de mango", 6500, 0.05),
    "006": ("Agua embotellada 500ml", 3000, 0.0),
}

try:
    with open("consecutivo.txt") as f:
        numero_factura = int(f.read())
except FileNotFoundError:
    numero_factura = 0

items = []

while True:
    print("\n1. Agregar producto  2. Ver factura  3. Cerrar factura  4. Salir")
    opcion = input("Opción: ")

    if opcion == "1":
        for cod, (nombre, precio, iva) in CATALOGO.items():
            print(cod, nombre, "$" + str(precio), "IVA", int(iva * 100), "%")
        codigo = input("Código: ")
        if codigo not in CATALOGO:
            print("Producto no encontrado")
            continue
        cantidad = int(input("Cantidad: "))
        nombre, precio, iva = CATALOGO[codigo]
        subtotal = precio * cantidad
        if cantidad > 3:
            subtotal = subtotal * 0.95  # Regla 2: desc. 5% por cantidad
        items.append((nombre, cantidad, subtotal, subtotal * iva))
        print("Agregado:", cantidad, nombre)

    elif opcion == "2" or opcion == "3":
        subtotal_total = sum(i[2] for i in items)
        iva_total = sum(i[3] for i in items)
        descuento = subtotal_total * 0.10 if subtotal_total > 30000 else 0  # Regla 1
        total = subtotal_total - descuento + iva_total

        print("=" * 40)
        print("CAFETERÍA LA ESQUINA S.A.S.")
        for nombre, cantidad, subtotal, iva_item in items:
            print(cantidad, "x", nombre, "=", round(subtotal))
        print("Subtotal:", round(subtotal_total))
        print("Descuento:", round(descuento))
        print("IVA:", round(iva_total))
        print("TOTAL:", round(total))
        print("=" * 40)

        if opcion == "3":
            if not items:
                print("No hay productos para facturar")
                continue
            numero_factura += 1
            with open("consecutivo.txt", "w") as f:
                f.write(str(numero_factura))
            with open(f"factura_{numero_factura}.txt", "w") as f:
                f.write(f"FACTURA No. {numero_factura}\n")
                for nombre, cantidad, subtotal, iva_item in items:
                    f.write(f"{cantidad} x {nombre} = {round(subtotal)}\n")
                f.write(f"Subtotal: {round(subtotal_total)}\n")
                f.write(f"Descuento: {round(descuento)}\n")
                f.write(f"IVA: {round(iva_total)}\n")
                f.write(f"TOTAL: {round(total)}\n")
            print(f"Factura guardada como factura_{numero_factura}.txt")
            items = []

    elif opcion == "4":
        print("Gracias por comprar en Cafetería La Esquina")
        break

    else:
        print("Opción no válida")
```
