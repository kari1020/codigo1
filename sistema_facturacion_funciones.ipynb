## 🖥️ El menú de la caja

El menú **ya está escrito** y no calcula nada: solo pregunta y llama a tus funciones.
Si tus funciones quedaron bien, funciona sin cambiarle una línea. Ese es el punto de hoy.


````python
def menu():
    while True:
        print("\n=== PAPELERÍA EL LÁPIZ FELIZ ===")
        print("1. Facturar un pedido")
        print("2. Consultar inventario")
        print("3. Ver facturas emitidas")
        print("4. Salir")
        opcion = input("Elige una opción: ").strip()

        if opcion == "1":
            doc = input("Documento del cliente: ").strip()
            if doc not in clientes:
                print("⚠️ Ese cliente no existe.")
                continue
            tipo = clientes[doc]["tipo"]
            lineas = []
            while True:
                codigo = input("Código del producto (ENTER para cerrar la factura): ").strip()
                if codigo == "":
                    break
                if buscar_producto(codigo) is None:
                    print("⚠️ Código inexistente.")
                    continue
                cantidad = int(input("Cantidad: "))
                if not descontar_inventario(codigo, cantidad):
                    print("⚠️ No hay suficiente stock.")
                    continue
                lineas.append(calcular_linea(codigo, cantidad, tipo))

            if not lineas:
                print("Factura cancelada: no se agregó ningún producto.")
                continue

            factura = calcular_factura(lineas, doc)
            factura["cliente"] = clientes[doc]["nombre"]
            facturas.append(factura)
            imprimir_factura(factura)
            print("Archivo generado:", guardar_factura_txt(factura))

        elif opcion == "2":
            consultar_inventario()

        elif opcion == "3":
            if not facturas:
                print("Todavía no se ha emitido ninguna factura.")
            for f in facturas:
                print("N.", f["numero"], "-", f["cliente"], "- $", round(f["total"]))

        elif opcion == "4":
            print("¡Hasta luego! 👋")
            break

        else:
            print("Opción no válida.")


# Ejecuta el menú cuando la revisión esté en verde:
# menu()
````
